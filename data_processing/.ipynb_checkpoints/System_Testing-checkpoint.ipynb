```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import json
import re
import seaborn as sns
from datetime import datetime
from shutil import move
import plotly.express as px
import pytz
import plotly.graph_objects as go

import glob
```

```python
df_mouse = pd.read_csv("/Users/shehjarsadhu/Desktop/UniversityOfRhodeIsland/Graduate/WBL/Project_MindGame/DATASET/Systems_testing_2025/HOME/N7777/Morning/session_2025-02-17_1/mouse_data/mouse_N7777_L4_S2_1739825693.csv")

```

```python


# Define the folder path
folder_path = "/Users/shehjarsadhu/Desktop/UniversityOfRhodeIsland/Graduate/WBL/Project_MindGame/DATASET/Systems_testing_2025/HOME/N7777/Morning/session_2025-02-17_1/mouse_data/"
average_sr_mouse_list = []
level_list_mouse = []
# Find all CSV files in the folder.
csv_files = glob.glob(f"{folder_path}/*.csv")
for csv_file in csv_files:
    df = pd.read_csv(csv_file)
    print("Puzzle Level: ",csv_file.split("/")[-1].split("_"))
    df["timestamp_seconds"] = df["timestamp"].apply(convert_timestamp)
    average_sample_rate = df["timestamp_seconds"].value_counts().mean()
    print("Average sample rate: ",average_sample_rate)
    average_sr_mouse_list.append(int(average_sample_rate))
    level_list_mouse.append(csv_file.split("/")[-1].split("_")[2] +"_"+ csv_file.split("/")[-1].split("_")[3])
# Use the hovertext kw argument for hover text
fig = go.Figure(data=[go.Bar(x=level_list_mouse, y=average_sr_mouse_list,            
                        text=average_sr_mouse_list,
                        textposition='auto',textfont=dict(size=20)  # Increase font size of text labels
)])
# Customize aspect
fig.update_traces(marker_color='rgb(158,202,225)', marker_line_color='rgb(8,48,107)',
                  marker_line_width=1.5, opacity=0.6)
# fig.update_layout(title_text='Average sample rate of computer mouse data')
# Update layout
fig.update_layout(
 title=dict(
        text='Average sample rate of computer mouse data',
        x=0.5,  # Center the title
        font=dict(size=24)  # Increase title font size
    ),
    xaxis=dict(
        tickfont=dict(size=22)  # Increase X-axis font size
    ),
    yaxis=dict(
        tickfont=dict(size=22)  # Increase Y-axis font size
    ),
    plot_bgcolor='white'  # Set background color to white
)
fig.show()
```

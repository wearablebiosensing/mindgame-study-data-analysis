## Statistical tests:

```python

# Function to analyze each variable
def analyze_variable(variable, in_lab_data, at_home_data):
    # Extract data
    x_in_lab = in_lab_data[variable]
    x_at_home = at_home_data[variable]

    # Normality Test
    shapiro_in_lab = shapiro(x_in_lab)
    shapiro_at_home = shapiro(x_at_home)
    
    # Variance Homogeneity Test
    levene_test = levene(x_in_lab, x_at_home)
    
    # Choosing the appropriate test
    if shapiro_in_lab.pvalue > 0.05 and shapiro_at_home.pvalue > 0.05 and levene_test.pvalue > 0.05:
        # Parametric test
        test_result = ttest_ind(x_in_lab, x_at_home)
        test_type = "t-test (Parametric)"
    else:
        # Non-parametric test
        test_result = mannwhitneyu(x_in_lab, x_at_home)
        test_type = "Mann-Whitney U test (Non-Parametric)"
    
    return {
        "Variable": variable,
        "Shapiro In-Lab p": shapiro_in_lab.pvalue,
        "Shapiro At-Home p": shapiro_at_home.pvalue,
        "Levene p": levene_test.pvalue,
        "Test Type": test_type,
        "Test Statistic": test_result.statistic,
        "p-value": test_result.pvalue
    }
```

```python
df_hr_cleaned = df_hr.dropna(subset=['scr_value_list'])
```

```python
def participant_wise_stats_hr(df_hr_cleaned,str_pid_column_name):
    pid_list = []
    variable_list = []
    p_value_list = []
    results_all = []
    for i in df_hr_cleaned[str_pid_column_name].unique():
        pid_subset = df_hr_cleaned[df_hr_cleaned[str_pid_column_name]==i]
        
        # Separate the data into in-lab (1) and at-home (0).
        in_lab = pid_subset[pid_subset['class_lab_0_home_1'] == 1]
        at_home = pid_subset[pid_subset['class_lab_0_home_1'] == 0]

        # Variables to analyze.
        variables = ["scr_value_list","percentage_missing_list_hr",'snr_hr_list_hr', 'percentage_0s_list_HR']

        # Apply the analysis for each variable.
        results = [analyze_variable(var, in_lab, at_home) for var in variables]
        results_all.append(results)
        # Display the results.
        for result in results:
            pid_list.append(i)
            print(f"  Variable: {result['Variable']}")
            variable_list.append(result['Variable'])
            p_value_list.append(result['p-value'])
            print(f"  Shapiro-Wilk Test (In-Lab) p-value: {result['Shapiro In-Lab p']:.4e}")
            print(f"  Shapiro-Wilk Test (At-Home) p-value: {result['Shapiro At-Home p']:.4e}")
            print(f"  Levene's Test p-value: {result['Levene p']:.4e}")
            print(f"  Test Used: {result['Test Type']}")
            print(f"  Test Statistic: {result['Test Statistic']:.4f}")
            print(f"  p-value: {result['p-value']}\n")
    df = pd.DataFrame(list(zip(pid_list, variable_list, p_value_list)), columns=['pid_list', 'variable_list', 'p_value_list'])
    return df
```

```python
def participant_wise_stats_mouse(df_mouse,str_pid_column_name):
    pid_list = []
    variable_list = []
    p_value_list = []
    for i in df_mouse[str_pid_column_name].unique():
        print("pid===============", i)
        pid_subset = df_mouse[df_mouse[str_pid_column_name]==i]
        # Separate the data into in-lab (1) and at-home (0)
        in_lab = pid_subset[pid_subset['class_lab_0_home_1'] == 1]
        at_home = pid_subset[pid_subset['class_lab_0_home_1'] == 0]
        print("in_lab: ",in_lab.columns)
        print("at_home: ",at_home.columns)
        # Variables to analyze
        variables = ["perentage_missing_non_negative", 'snr_mouse_x_list','snr_mouse_y_list']

        # Apply the analysis for each variable
        results = [analyze_variable(var, in_lab, at_home) for var in variables]

        # Display the results
        for result in results:
            pid_list.append(i)
            variable_list.append(result['Variable'])
            p_value_list.append(result['p-value'])
            print(f"Variable: {result['Variable']}")
            print(f"  Shapiro-Wilk Test (In-Lab) p-value: {result['Shapiro In-Lab p']:.4e}")
            print(f"  Shapiro-Wilk Test (At-Home) p-value: {result['Shapiro At-Home p']:.4e}")
            print(f"  Levene's Test p-value: {result['Levene p']:.4e}")
            print(f"  Test Used: {result['Test Type']}")
            print(f"  Test Statistic: {result['Test Statistic']:.4f}")
            print(f"  p-value: {result['p-value']}\n")


    df = pd.DataFrame(list(zip(pid_list, variable_list, p_value_list)), columns=['pid_list', 'variable_list', 'p_value_list'])
    return df
```
